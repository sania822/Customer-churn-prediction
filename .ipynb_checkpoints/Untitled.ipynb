```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

#that data define we have 
train_df = pd.read_csv("churn-bigml-80.csv")
test_df = pd.read_csv("churn-bigml-20.csv")

#head dataset
test_df.head()
```

```python
#map the churn column into binary
train_df['Churn'] = train_df['Churn'].map({True:1, False:0})
test_df['Churn'] = test_df['Churn'].map({True:1, False:0})
```

```python
train_df['Churn'].head()
```

```python
#Align columns
train_df , test_df = train_df.align(test_df , join = "left", axis = 1 , fill_value = 0)
```

```python
#dropping columns
train_df = train_df.drop(["State", "Area code"], axis=1)
```

```python
train_df.head()
```

****Feature selection****

```python

train_df.rename(columns = {"International plan_Yes": "intenational_plan", 
                          "Voice mail plan_Yes": "voice_mail_plan"}, 
               inplace=True)  
```

```python
# split features and target
X = train_df.drop("Churn" , axis = 1)
y = train_df["Churn"]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
x_scaled = pd.DataFrame(X_scaled , columns = X.columns)
```

```python
X_scaled
```

```python
from sklearn.model_selection import train_test_split

X_train , X_val , y_train , y_val = train_test_split(
                                               X_scaled, y,
                                                test_size=0.2,
                                                random_state=42,
                                                stratify=y
                                            )
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(class_weight="balanced")
model.fit(X_train, y_train)
```

```python
#check how good the model was

y_pred = model.predict(X_val)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_val, y_pred)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_val, y_pred)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_val, y_pred))
```

```python
#Which factors increase churn probability most?
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.coef_[0]
})

feature_importance.sort_values("Importance", ascending=False)
```

```python
train_df.rename(columns = {"Voice mail plan_Yes" : "voice_mail_plan" , "International plan_Yes" :"international_plan"}).inplace = True
```

```python
#make model more powerful 

from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(
    n_estimators = 100,
    class_weight = "balanced",
    random_state = 42)
```

```python
rfc.fit(X_train , y_train)
```

```python
rf_pred = rfc.predict(X_val)
```

```python
print(classification_report(y_val, rf_pred))
```

*****Hyperparameter tuning*****

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "n_estimators" : [100,200],
    "max_depth" : [None , 10 , 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}
```

```python
grid_search = GridSearchCV(
    RandomForestClassifier(class_weight="balanced", random_state=42),
    param_grid,
    cv=5, #5-fold cross validation
    scoring="recall", #maximize churn detection
    n_jobs=-1  #faster (uses all CPU)
)
```

```python
#train the model with grid search
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:", grid_search.best_params_)
```

```python
best_rf = grid_search.best_estimator_
```

```python
#predict the validation set with the best model
best_pred = best_rf.predict(X_val)
```

```python
#classification report 
print(classification_report(y_val, best_pred))
```
